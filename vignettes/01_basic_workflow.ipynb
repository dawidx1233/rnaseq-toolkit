# Basic RNA-seq Workflow with `rnaseq-toolkit`



````python
import pandas as pd
from rnaseq_toolkit.pipeline import RNAseqPipeline

# Load example data (replace with your own files)
counts = pd.read_csv('../data/example_counts.csv', index_col=0)
metadata = pd.read_csv('../data/example_metadata.csv', index_col=0)

print(f"Counts: {counts.shape[0]} genes x {counts.shape[1]} samples")
````

## Initialize and Run the Pipeline



````python
pipeline = RNAseqPipeline(
    counts=counts,
    metadata=metadata,
    design='~condition',
    contrast=['condition', 'Treated', 'Control'],
    norm_method='deseq2',
    dea_method='deseq2',
    output_dir='results_basic/'
)

# Run the full pipeline (Normalization, DEA, Plots, HTML Report)
````

## View Results

The pipeline automatically saves normalized counts, DEG tables, and high-quality plots in the `results_basic/` directory.

```python
# You can also access the results programmatically:
res = pipeline.results
significant_degs = res[(res['padj'] < 0.05) & (res['log2FoldChange'].abs() > 1)]
